## (*) Triangular numbers

image.png




The triangular number T<sub>n</sub> solves the handshake problem of counting the number of handshakes if each person in a room with n + 1 people shakes hands once with each person. In other words, the solution to the handshake problem of n people is T<sub>n−1</sub>.

T<sub>1</sub> = 1

T<sub>2</sub> = 1 + 2 = 3

T<sub>3</sub> = 1 + 2 + 3 = 6

T<sub>4</sub> = 1 + 2 + 3 + 4 = 10

T<sub>5</sub> = 1 + 2 + 3 + 4 + 5 = 15


Write a recursive program to solve for T<sub>n</sub>.


````python
def triangular_term(n):
    return n if n == 1 else n+triangular_term(n-1)

    """
    Calculates the sum of all integers from 1 to the given number n.

    Parameters:
    n (int): The number up to which to calculate the sum.

    Returns:
    int: The sum of integers from 1 to n.

    Examples:
    >>> triangular_term(5)
    15
    >>> triangular_term(10)
    55
    >>> triangular_term(100)
    5050
    """

triangular_term(100)
````

## (*) Fibonacci Number

image.png

````python
def fib(n):
    return n if n <= 1 else fib(n-1) + fib(n-2)
    '''
    Calculates the nth fibonacci number
    >>> fib(0)
    0

    >>> fib(1)
    1

    >>> fib(5)
    5

    >>> fib(10)
    55
    '''

fib(10)
````

Python provides a built-in function called `count()` that can be used to count the number of occurrences of a substring or character in a string. Here's an example of how to use it:

```python
s = "hello world"
char = "l"
count = s.count(char)
print(count)  # Output: 3
```

````python
def count(s, ch, n=0):
    return (n 
            if len(s) == 0 
            else (
                count(s[1:], ch, n=n+1) 
                if s[0] == ch 
                else count(s[1:], ch, n=n)
                )
            )

count("strawberry", "r")


# Parameters:
# s (str): The input string.
# h (str): The character to count occurrences of.

# Returns:
# int: The number of times the character appears in the string.
````
