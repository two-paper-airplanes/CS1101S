# Source Academy Python: Curves and Sound

These notes summarise the main ideas, functions, formulas, and common mistakes for curves and sound in Source Academy Python.

> These examples are written for the Source Academy environment. Some functions, such as `make_point` and `make_sound`, are not standard Python functions.

````python
# fake imports
import make_point, draw_connected, make_sound, play from source-academy
import math
````

## 1. Curves

A curve is a function that takes a parameter `t`, usually from `0` to `1`, and returns a point.

```python
curve = lambda t: make_point(x_expression, y_expression)
```

The curve is sampled at different values of `t` and joined together to approximate the shape.

````python
line = lambda t: make_point(t, t)
horizontal = lambda t: make_point(t, 0.5)
vertical = lambda t: make_point(0.5, t)

draw_connected(200, line)
````

### Circle

The standard parameterisation of a circle uses sine and cosine:

`x = centre_x + radius * cos(2πt)`

`y = centre_y + radius * sin(2πt)`

````python
circle = lambda t: make_point(
    0.5 + 0.5 * math.cos(2 * math.pi * t),
    0.5 + 0.5 * math.sin(2 * math.pi * t)
)

draw_connected(300, circle)
````

### Curve transformations

- **Scale:** multiply the x- or y-coordinate.
- **Translate:** add a constant to the x- or y-coordinate.
- **Reverse:** use `curve(1 - t)`.
- **Connect:** split `[0, 1]` into intervals and use a different curve on each interval.

```python
reversed_curve = lambda t: curve(1 - t)

connected = lambda t: (
    c1(2 * t) if t <= 0.5
    else c2(2 * t - 1)
)
```

## 2. Sound

A sound is produced from a wave function. The wave function takes time `t` in seconds and returns the amplitude at that moment.

```python
make_sound(wave, duration)
```

````python
def sine_sound(freq, duration):
    wave = lambda t: math.sin(2 * math.pi * freq * t)
    return make_sound(wave, duration)

tone = sine_sound(800, 1)
play(tone)
````

### Sine-wave formula

`sin(2πft)`

- `f` is the frequency in Hertz.
- `t` is time in seconds.
- A higher frequency produces a higher pitch.
- `math_sin` uses radians, so the `2 * math_pi` factor is needed.

## 3. Combining sounds

### Consecutive sounds

`consecutively(sound1, sound2)` plays `sound2` after `sound1` ends.

For three sounds:

```python
consecutively(sound1, consecutively(sound2, sound3))
```

### Simultaneous sounds

`simultaneously(sound1, sound2)` starts both sounds at the same time. This can be used to create chords.

## 4. Cutting sound

To keep only part of a sound, use:

```python
cut_sound(sound, start, duration)
```

For example, `cut_sound(sound, 0.5, 1)` starts at 0.5 seconds and keeps the next 1 second.

## 5. Morse-code timing

For the usual Source Academy SOS exercise:

| Element | Duration |
|---|---:|
| Dot | `0.125` s |
| Dash | `3 * 0.125 = 0.375` s |
| Silence within a letter | `0.125` s |
| Silence between letters | `0.375` s |

```python
dot_duration = 0.125
dash_duration = 3 * dot_duration
dot = sine_sound(800, dot_duration)
dash = sine_sound(800, dash_duration)
silence_dot = silence_sound(dot_duration)
silence_dash = silence_sound(dash_duration)
```

`S` is dot-dot-dot, and `O` is dash-dash-dash. Join the sounds with `consecutively`. Do not add pauses before or after the complete signal if the question says not to.

## 6. Common mistakes

1. A curve must return a point, not just a number.
2. Remember that `t` normally runs from `0` to `1`.
3. Frequency is measured in Hertz; duration is measured in seconds.
4. `math_sin` and `math_cos` use radians.
5. Use `consecutively` for sounds one after another and `simultaneously` for overlapping sounds.
6. Check the exact required duration and silence timing in Morse-code questions.
