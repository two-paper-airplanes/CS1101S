### Exercise 1.29



```python
cube = lambda x: x*x*x
n = 100

h = (1-0)/n

def coeff(k):
    if k == 0 or k == n:
        return 1
    return 4 if k % 2 == 1 else 2

y_k = lambda f: lambda k: coeff(k) * f(0 + k*h)

sum = 0
for j in range(n+1):
    sum = sum+ y_k(cube)(j)

print((h / 3)* sum)
```
