```python
# Question 17
# What kind of process does the function fun give rise to for any integer argument x > 0?

def fun(x):
    f = lambda x: x + 1
    return f(x) if f(x) < 2 else f(fun(x - 1))

# lets try with example
# fun(4)                                                -> f(4) = 5 ≥ 2
# f(fun(3))                                             -> f(3) = 4 ≥ 2
# f(f(fun(2)))                                          -> f(2) = 3 ≥ 2
# f(f(f(fun(1))))                                       -> f(1) = 2 ≥ 2
# f(f(f(f(fun(0)))))                                    -> f(0) = 1 < 2   ->   return 1
# f(f(f(f(1)))) = f(f(f(2))) = f(f(3)) = f(4) = 5       -> since there are deferred operations this is recursive !!
```
